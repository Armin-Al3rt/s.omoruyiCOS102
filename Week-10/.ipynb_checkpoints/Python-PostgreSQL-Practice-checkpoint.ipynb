# Connecting to a Database

```python
import psycopg2

# Database connection parameters
host = "localhost"
port = 5432
dbname = "globacom_dbase"
user = "postgres"
password = "ELLEN_IMAOBONG"

try:
    connection = psycopg2.connect(
        host=host,
        port=port,
        dbname=dbname,
        user=user,
        password=password
    )
    print("Connection to PostgreSQL DB successful")

    cursor = connection.cursor()

    # Correct SQL query to get PostgreSQL version
    cursor.execute("SELECT version();")
    
    db_version = cursor.fetchone()
    print(f"PostgreSQL database version: {db_version[0]}")

    cursor.close()
    connection.close()

except Exception as error:
    print(f"Error connecting to PostgreSQL DB: {error}")
```

# Selecting existing Database

```python
import psycopg2

# Database connection parameters
host = "localhost"
port = 5432
dbname = "globacom_dbase"
user = "postgres"
password = "ELLEN_IMAOBONG"

try:
    connection = psycopg2.connect(
        host=host,
        port=port,
        dbname=dbname,
        user=user,
        password=password
    )
    print("Connection to PostgreSQL DB successful")

    cursor = connection.cursor()

    # Correct SQL query selecting all columns
    select_query = "SELECT * FROM staff;"

    cursor.execute(select_query)

    data = cursor.fetchall()

    # If you want to print column names dynamically:
    colnames = [desc[0] for desc in cursor.description]
    print(" | ".join(colnames))
    print("-" * (len(" | ".join(colnames)) + 5))

    for row in data:
        print(" | ".join(str(item) for item in row))

    cursor.close()
    connection.close()

except Exception as error:
    print(f"Error connecting to PostgreSQL DB: {error}")
```

# Create a new Database

```python
import psycopg2
from psycopg2 import sql

# Parameters to connect to the existing PostgreSQL server
host = "localhost"
port = 5432
# Usually connect to default "postgres" database to create others
dbname = "postgres"  
user = "postgres"
password = "ELLEN_IMAOBONG"

# Name of the new database to be created
new_dbname = "cos102_db"

try:
    # Connect to the PostgreSQL server (default database)
    connection = psycopg2.connect(
        host=host,
        port=port,
        dbname=dbname,
        user=user,
        password=password
    )
    connection.autocommit = True  # Enable autocommit mode
    print("Connection to PostgreSQL server successful")

    cursor = connection.cursor()

    # Define SQL to create new database safely using sql.Identifier
    create_db_query = sql.SQL("CREATE DATABASE {}").format(sql.Identifier(new_dbname))

    cursor.execute(create_db_query)
    print(f"Database '{new_dbname}' created successfully")

    cursor.close()
    connection.close()

except Exception as error:
    print(f"Error creating the database: {error}")
```

# Create a Table

```python
import psycopg2

# Database connection parameters
host = "localhost"
port = 5432
dbname = "globacom_dbase"
user = "postgres"
password = "ELLEN_IMAOBONG"

try:
    # Connect to the PostgreSQL database
    connection = psycopg2.connect(
        host=host,
        port=port,
        dbname=dbname,
        user=user,
        password=password
    )
    # connection.autocommit = True  # Not necessary for CREATE TABLE inside transaction
    
    print("Connection to PostgreSQL DB successful")

    # Create a cursor object
    cursor = connection.cursor()  # fixed variable name

    # Define the SQL statement to create a new table
    create_table_query = """
    CREATE TABLE sst_info (
        id SERIAL PRIMARY KEY,
        name VARCHAR(50) NOT NULL,
        age INT NOT NULL,
        department VARCHAR(50) NOT NULL,
        password VARCHAR(50) NOT NULL
    );
    """

    # Execute the SQL statement
    cursor.execute(create_table_query)
    connection.commit()  # Commit the changes
    print("Table created successfully")

    # Close the cursor and connection
    cursor.close()
    connection.close()

except Exception as error:
    print(f"Error creating table: {error}")

```

# INSERT into Table

```python
import psycopg2

# Database connection parameters
host = "localhost"
port = 5432
dbname = "globacom_dbase"
user = "postgres"
password = "ELLEN_IMAOBONG"

try:
    # Connect to the PostgreSQL database
    connection = psycopg2.connect(
        host=host,
        port=port,
        dbname=dbname,
        user=user,
        password=password
    )
    connection.autocommit = True  # Enable autocommit mode
    print("Connection to PostgreSQL DB successful")

    # Create a cursor object
    cursor = connection.cursor()

    # Define the SQL INSERT statement (exclude id to auto-generate)
    insert_query = """
    INSERT INTO sst_info (name, age, department, password)
    VALUES (%s, %s, %s, %s);
    """
    
    # Data to be inserted
    data_to_insert = ("Jamel Ojo", 20, "Software Engineering", "anthelo4")

    # Execute the SQL statement
    cursor.execute(insert_query, data_to_insert)
    print("Data inserted successfully")

    # Close the cursor and connection
    cursor.close()
    connection.close()

except Exception as error:
    print(f"Error inserting data: {error}")
```

# SELECT from Table

```python
import psycopg2

# Define your database connection parameters
db_params = {
    "host": "localhost",
    "port": 5432,
    "dbname": "globacom_dbase",
    "user": "postgres",
    "password": "ELLEN_IMAOBONG"
}

try:
    # Establish a connection to the database
    connection = psycopg2.connect(**db_params)

    # Create a cursor object
    cursor = connection.cursor()

    # Define the SQL query (select all columns where age < 30)
    select_query = "SELECT * FROM sst_info WHERE age < 30;"

    # Execute the SQL query
    cursor.execute(select_query)

    # Fetch all the results
    results = cursor.fetchall()

    # Print the results
    for row in results:
        print(row)

except Exception as error:
    print(f"Error while connecting to PostgreSQL: {error}")

finally:
    # Close the cursor and connection to clean up
    try:
        if cursor:
            cursor.close()
    except NameError:
        pass

    try:
        if connection:
            connection.close()
    except NameError:
        pass
```
